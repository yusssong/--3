
# langchain
## 1langchain介绍
<br/>
LangChain 是一个开源框架(一套在大模型能力上封装的工具框架)，它旨在帮助开发者轻松构建应用程序，这些应用程序能够利用语言模型（特别是大型语言模型）来实现各种功能。LangChain 的核心理念是提供一个结构化的方法来连接不同的组件（如语言模型、数据源等），以便构建复杂的应用程序。<br/>
(langchain是帮助我们调用大语言模型) langchain把和大型语言模型的交互的流程抽象出来，变成一个一个的小模块，方便更好的进行调用。通过把这些小模块串联起来，实现更多更复杂的功能。

### 模块
<br/>
LangChain提供了对几个主要模块的支持。针对每个模块，后续的文章介绍了一些基础知识并提供了相应的示例。这些模块按照复杂性排列如下：<br/>
模型（Models）：LangChain支持的各种模型类型和模型集成。<br/>
提示（Prompts）：包括提示管理、提示优化和提示序列化。<br/>
记忆（Memory）：记忆是在链或代理调用之间保持状态的组件，LangChain提供了一个标准的记忆接口。<br/>
索引（Indexes）：与我们自己的文本数据结合使用时，语言模型往往更加强大，此模块涵盖了执行此操作的最佳实践。<br/>
链（Chains）：链不仅仅是单个LLM的调用，还包括了一系列的其它调用（调用LLM或不同的实用工具等）。LangChain提供了一种标准的链接口和许多与其他工具的集成，还提供了用于常见应用程序的端到端的链调用。<br/>
代理（Agents）：代理涉及LLM做出行动决策、执行该行动、查看一个观察结果等过程，并重复该过程直到完成。LangChain提供了一个标准的代理接口，其包含了一系列可供选择的代理以及端到端代理的示例。<br/>

### 用例
上述模块可以以多种方式使用。LangChain提供了指导和帮助。以下是LangChain支持的一些常见用例。<br/>
自治代理（Autonomous Agents）：长时间运行的代理会采取多步操作以尝试完成目标。AutoGPT和BabyAGI就是典型代表。<br/>
代理模拟（Agent Simulations）：将代理置于封闭环境中观察它们如何相互作用，如何对事件作出反应，是观察它们长期记忆能力的有趣方法。<br/>
个人助理（Personal Assistants）：主要的LangChain使用用例。个人助理需要采取行动、记住交互并具有您的有关数据的知识。<br/>
问答（Question Answering）：第二个重大的LangChain使用用例。仅利用这些文档中的信息来构建答案，回答特定文档中的问题。<br/>
聊天机器人（Chatbots）：由于语言模型擅长生成文本，因此它们非常适合创建聊天机器人。<br/>
查询表格数据（Tabular）：使用LLM查询存储在表格格式中的数据（csv、SQL、数据框等）。<br/>
代码理解（Code）：使用LLM查询来自GitHub的源代码。<br/>
与API交互（APIs）：使LLM能够与API进行交互，以便为它们提供更实时的信息并允许它们采取行动。<br/>
提取（Extraction）：从文本中提取结构化信息。<br/>
摘要（Summarization）：将较长的文档汇总为更短、更简洁的信息块，是一种数据增强生成的类型。<br/>
评估（Evaluation）：生成模型是极难用传统度量方法评估的。 一种新的评估方式是使用语言模型本身进行评估。LangChain提供一些用于辅助评估的提示或链。<br/>

## 3.提示词（Prompts）
<br>
语言模型以文本作为输入，这段文本通常被称为提示（Prompt）。通常情况下，这不仅仅是一个硬编码的字符串，而是模板、示例和用户输入的组合。LangChain提供了多个类和函数，以便轻松构建和处理提示。<br>
你可以理解 Prompt是与LLM沟通的方式，Prompt编写的好坏决定了最后的效果，关于Prompt也有提示词是一门大学问。<br/>

### 提示词一般有以下要素组成：<br/>
角色：给大模型定义一个匹配目标任务的角色。用一句话就可以明确它的角色（比如“我是一位资深java工程师”），从而有效的收窄问题域，减少二义性，让“通用”瞬间变得“专业”。<br/>

指示：对具体任务进行详细描述。<br/>

上下文：给出与任务相关的其它背景信息（如历史对话、情境等）。<br/>

例子：举例很重要，就像是师傅教学之后，需要给徒弟（大模型）演示一下如何操作，这个手把手的操作，是大模型生成输出时的一个重要参考，对输出结果有很大帮助。<br/>

输入：任务的输入信息，最好在提示词中有明确的“输入”标识。<br/>

输出：输出的格式描述，比如用郭德纲的语气、输出不超过十个字、以JSON格式返回结果等。<br/>

提示词的使用技巧: 清晰，明确，避免模糊的词语。

### 提示词的的构建四种消息类型 SystemMessage,HumanMessage,AiMessage,ToolMessage


System：系统消息
用来规定模型的身份、规则、风格和边界。它优先级最高，通常由开发者设置。

{"role": "system", "content": "你是一个耐心的物理老师，用初中生能听懂的话解释。"}

```python
from langchain.messages import SystemMessage

system_message = SystemMessage(content="你是一个耐心的物理老师，用初中生能听懂的话解释。")
print(system_message)
print(type(system_message))
```

Human/user：用户消息
表示用户真正提出的问题或任务。

{"role": "human", "content": "什么是量子计算？"}

```python
from langchain.messages import HumanMessage

human_message = HumanMessage(content="什么是量子计算？")
print(human_message)
print(type(human_message))
```

Ai：模型消息
表示模型之前已经回复过的内容。多轮对话时，把历史回答放进去，模型才能“记得”上下文<br>
{"role": "ai", "content": "量子计算是一种利用量子比特进行计算的方法。"}

Tool：工具消息
表示外部工具返回的结果，比如搜索、数据库查询、计算器、天气 API。模型不能自己假装工具结果，应该根据真实工具返回继续回答<br>

{"role": "tool", "content": "北京今天气温 28℃，多云。"}


当你想要与一个语言模型进行交互时，你可以使用来SystemMessage,HumanMessage构造你的提示词，您可以更好地控制模型的行为，例如设置模型的角色、语气等

```python
from langchain.messages import HumanMessage,SystemMessage,AIMessage,ToolMessage
from langchain_ollama import ChatOllama

from langchain_classic.globals import set_debug, set_verbose
# 启用调试模式
set_debug(True)
set_verbose(True)

llm = ChatOllama(model="qwen3:14b",base_url="http://192.168.1.105:11434")

# 创建一个系统消息，定义对话的背景或指导模型的行为

def test_llm(input_message):
    system_message = SystemMessage(content="你是物理老师。")
    messages = [
        system_message,
        HumanMessage(content=input_message)
    ] 
    response = llm.invoke(messages)
    return response

if __name__ == "__main__":
    input_message = "解释一下什么是量子计算？"
    response = test_llm(input_message)
    print(response)
```

### 提示模板(类似于开始过程中遇到的短信模板、邮件模板)
提示模板是指一种可复制的生成提示的方式。它包含一个文本字符串（模板），可以从最终用户处接收一组参数并生成提示。提示模板可能包含以下内容：<br/>

+ 对语言模型的指令<br/>
+ 提供一些简单示例给大语言模型从而使模型输出更接近理想结果<br/>
+ 提给语言模型的问题<br/>
+ 将提示词提炼成模板<br/>
+ 实现提示词的复用、版本管理、动态变化等<br/>

ChatPromptTemplate：聊天消息模板，适用于多轮对话

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import SystemMessage,HumanMessage
from langchain_ollama import ChatOllama
from langchain_classic.globals import set_debug, set_verbose
# 启用调试模式
set_debug(True)
set_verbose(True)

llm = ChatOllama(model="qwen2.5:7b",url="http://localhost:11434")

prompt = ChatPromptTemplate.from_messages([
    ("system", "你是物理{system_name}老师。"),
    ("human", "解释一下什么是量子计算,你好{name}?")]
)  #需要去区分角色

print(type(prompt))

# 调用 qwen 模型，传入消息列表
messages = prompt.format_messages(system_name="高级",name="张三")
print(messages)
print(type(messages))


print("--------------------------------------------------------------------------------------")


prompt2 = ChatPromptTemplate.from_messages([
    ("system", "你是物理{system_name}老师。"),
    ("human", "解释一下什么是量子计算,你好{name}?")
])  


messages2 = prompt2.format_prompt(system_name="高级",name="张三")
print(messages2)
print(type(messages2))

print("--------------------------------------------------------------------------------------")

print(messages2.to_messages())
print(type(messages2.to_messages()))
print("--------------------------------------------------------------------------------------")

print(messages2.to_string())
print(type(messages2.to_string()))
print("--------------------------------------------------------------------------------------")


'''
format_messages(...) → List[BaseMessage]
输出：结构化“消息列表”，如 [SystemMessage, HumanMessage, ...]
适用：下游接口需要逐条消息（多轮/带角色）的输入
优点：保留角色与元数据，便于聊天模型准确解析

format_prompt/invoke 两种方式得到的变量类型都是(...) → PromptValue
输出：中间对象 PromptValue
用法：既可 .to_messages() 得到消息列表，也可 .to_string() 得到纯文本
适用：需要灵活在“消息列表/字符串”之间切换
'''
```

### 提供示例给LLM从而实现逻辑思维(Few-shot prompt)

FewShotPromptTemplate 示例


```python
from langchain_core.prompts.few_shot import FewShotPromptTemplate, FewShotChatMessagePromptTemplate
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
from langchain_ollama import ChatOllama

# 初始化本地LLM模型
llm = ChatOllama(model="qwen3:14b",base_url="http://192.168.1.105:11434")

# 定义示例数据：包含问题和答案的字典列表
examples = [
    {
        "question": "你好吗?",
        "answer": "师哥，我很好"
    },
    {
        "question": "今天周几?",
        "answer": "师哥，今天星期二"
    },
    {
        "question": "天气好吗?",
        "answer": "师哥，是的，今天天气很好"
    }
]

# 创建示例模板：定义如何格式化每个示例
example_prompt = PromptTemplate(
    input_variables=["question", "answer"],  # 模板中使用的变量
    template="问题: {question}\n回答: {answer}\n"  # 模板格式
)

# 创建少样本提示词模板
prompt = FewShotPromptTemplate(
    examples=examples,                    # 示例列表
    example_prompt=example_prompt,        # 示例格式化模板
    prefix="你是一个礼貌的助手，总是以'师哥'开头回答问题。以下是示例：\n",  # 提示词前缀
    suffix="问题: {input}\n回答:",        # 提示词后缀，包含用户输入
    input_variables=["input"]             # 输入变量
)

# 格式化提示词并打印
print(prompt.format(input="明天星期几?"))
```

FewShotChatMessagePromptTemplate 示例用于聊天模型的少样本学习提示模板

```python
from langchain_core.prompts import (
    ChatPromptTemplate,
    FewShotChatMessagePromptTemplate,
)
from langchain_ollama import ChatOllama

# 初始化模型
llm = ChatOllama(model="qwen2.5:7b",base_url="http://localhost:11434")

# 定义少样本示例
examples = [
    {"question": "你好吗?", "answer": "师哥，我很好"},
    {"question": "今天周几?", "answer": "师哥，今天星期二"},
    {"question": "天气好吗?", "answer": "师哥，是的，今天天气很好"}
]

# 定义示例格式化的提示模板
# 使用 ChatPromptTemplate 来定义每个示例的格式
example_prompt = ChatPromptTemplate.from_messages(
    [
        ("human", "{question}"),
        ("ai", "{answer}"),
    ]
)

# 创建 FewShotChatMessagePromptTemplate
few_shot_prompt = FewShotChatMessagePromptTemplate(
    example_prompt=example_prompt,
    examples=examples,
)


print(few_shot_prompt)
# 创建完整的提示模板
final_prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "你是一个中二的服务员，请用类似的风格回答问题。"),
        few_shot_prompt,
        ("human", "{input}"),
    ]
)

formatted_messages = final_prompt.format_messages(input="明天星期几?")

print(formatted_messages)
# 调用模型
chain = final_prompt | llm
response = chain.invoke({"input": "明天星期几?"})
print("\n模型回答:")
print(response.content)
```

 高级示例：使用动态示例选择器

```python
# 导入示例选择器类
from langchain_core.example_selectors import SemanticSimilarityExampleSelector  

from langchain_community.vectorstores import FAISS #向量数据库
from langchain_ollama import OllamaEmbeddings #编码模型
from langchain_core.prompts import PromptTemplate,FewShotPromptTemplate

# 扩展的示例数据集
more_examples = [
    {"question": "你好吗?", "answer": "师哥，我很好"},
    {"question": "今天周几?", "answer": "师哥，今天星期二"},
    {"question": "天气好吗?", "answer": "师哥，是的，今天天气很好"},
    {"question": "你叫什么?", "answer": "师哥，我是AI助手"},
    {"question": "几点了?", "answer": "师哥，我无法获取当前时间"},
    {"question": "什么是卷积神经网络", "answer": "卷积神经网络就是一个滤波器"},
    {"question": "解释一下机器学习", "answer": "师哥，机器学习是人工智能的一个分支..."},
]

example_prompt = PromptTemplate(
    input_variables=["question", "answer"],  # 模板中使用的变量
    template="问题: {question}\n回答: {answer}\n"  # 模板格式
)

# 创建嵌入模型：将文本转换为向量
embeddings = OllamaEmbeddings(model="qwen3-embedding:0.6b",base_url="http://192.168.1.105:11434")  
# 创建语义相似性选择器
# 根据输入问题与示例问题的语义相似性来选择最相关的示例
semantic_selector = SemanticSimilarityExampleSelector.from_examples(
            examples=more_examples, 
            embeddings=embeddings,
            vectorstore_cls=FAISS,          # 向量存储
            k=2                         # 选择最相似的2个示例

        )
        
question = "什么是卷积神经网络？"
print(semantic_selector.select_examples({"question": question}))


# 使用语义选择器创建提示词模板
semantic_prompt = FewShotPromptTemplate(
            example_selector=semantic_selector,
            example_prompt=example_prompt,
            prefix="基于相似问题，请回答：\n",
            suffix="问题: {question}\n回答:",
            input_variables=["question"]
        )
        
        # 测试语义选择器：输入与机器学习相关的问题
print(semantic_prompt.format(question="什么是卷积神经网络？"))
```

## 4.链（chain）

在许多编程语言和库中，"chain"通常用于描述一系列的操作或函数，这些操作或函数按照特定的顺序依次执行，前一个操作的输出会作为后一个操作的输入。这种模式也被称为管道（Pipeline）或链式调用（Chain Calling）。
(小的模块串联起,形成一种链式结构.这种模块化的编码方式非常有利于项目开发。)<br/>

<br/>

"chain"的作用如下：<br/>
+ 1.将处理流程串联起来：通过将一系列的操作串联起来，可以使处理流程更清晰，更易于理解和维护。<br/>
+ 2.简化代码：通过链式调用，可以避免在每一步操作后都需要单独保存和处理结果，从而简化代码。<br/>
+ 3.提高灵活性：通过改变链中的操作或其顺序，可以轻松地修改处理流程，以适应不同的需求。<br/>
+ 4.提高效率：在某些情况下，链式调用可以提高代码的执行效率。例如，如果链中的每个操作都是异步执行的，那么整个处理流程可能会比逐个执行这些操作更快。<br/>


langchain 使用LCEL表达式将将不同的模块链接到一起，使用管道符 | 表示，这种方式既方便书写，表达力也很强劲。<br/>



```python
from langchain_core.prompts import PromptTemplate
from langchain_ollama import ChatOllama

# 提示词
prompt = PromptTemplate(
    input_variables=["content1", 'content2'],
    template="关于{content1}{content2}是什么?请中文回复我"
)
# 初始化模型
llm = ChatOllama(model="qwen2.5:7b",base_url="http://localhost:11434")
# 构建一个链路
chain = prompt | llm  #LCH
# 运行
res = chain.invoke({'content1': '苹果', 'content2': '笑话'})
print(res)
```

##### Router chain(多链结构)（老版本中的内容了解即可）
<br>
**路由链（RouterChain）**是由LLM根据输入的Prompt去选择具体的某个链。路由链中一般会存在多个Prompt，Prompt结合LLM决定下一步选择哪个链。简单来说，就是根据“用户输入”的信息，能够动态的选择交给“谁”去解决问题.<br>

```python
from langchain_ollama import OllamaLLM

from langchain.chains.router import LLMRouterChain, MultiPromptChain
from langchain.chains.router.llm_router import RouterOutputParser
from langchain.chains.router.multi_prompt_prompt import MULTI_PROMPT_ROUTER_TEMPLATE
from langchain.chains import LLMChain, ConversationChain
from langchain.prompts import PromptTemplate

model = OllamaLLM(model="qwen2.5:7b")

# 准备3条目的链：一条物理链，一条数学链，一条英语链
# 1. 物理链
physics_template = """
你是一位物理学家，擅长回答物理相关的问题，当你不知道问题的答案时，你就回答不知道。
具体问题如下：
{input}
"""
physics_prompt = PromptTemplate.from_template(physics_template)
physics_chain = LLMChain(llm=model, prompt=physics_prompt)
#physics_chain = physics_prompt | model
# 2. 数学链
math_template = """
你是一个数学家，擅长回答数学相关的问题，当你不知道问题的答案时，你就回答不知道。
具体问题如下：
{input}
"""
math_prompt = PromptTemplate.from_template(math_template)
math_chain = LLMChain(llm=model, prompt=math_prompt)
#math_chain = math_prompt | model
# 3. 英语链
english_template = """
你是一个非常厉害的英语老师，擅长回答英语相关的问题，当你不知道问题的答案时，你就回答不知道。
具体问题如下：
{input}
"""
english_prompt = PromptTemplate.from_template(english_template)
english_chain = LLMChain(llm=model, prompt=english_prompt)
#english_chain = english_prompt | model
######### 所有可能的目的链
destination_chains = {}
destination_chains["physics"] = physics_chain
destination_chains["math"] = math_chain
destination_chains["english"] = english_chain
######### 默认链
default_chain = ConversationChain(llm=model, output_key="text")
# 让多路由模板 能找到合适的 提示词模板
destinations_template_str = """
physics:擅长回答物理问题
math:擅长回答数学问题
english:擅长回答英语问题
"""
router_template = MULTI_PROMPT_ROUTER_TEMPLATE.format(
    destinations=destinations_template_str
)
# 通过路由提示词模板，构建路由提示词
router_prompt = PromptTemplate(
    template=router_template,
    input_variables=["input"],
    output_parser=RouterOutputParser(),
)
######### 路由链
router_chain = LLMRouterChain.from_llm(llm=model, prompt=router_prompt)
######### 最终的链
multi_prompt_chain = MultiPromptChain(
    router_chain=router_chain,
    destination_chains=destination_chains,
    default_chain=default_chain,
    verbose=True,
)
res = multi_prompt_chain.invoke({"input": "重力加速度是多少？"})
# multi_prompt_chain.invoke("y=x^2+2x+1的导数是多少？")
# res = multi_prompt_chain.invoke("将以下英文翻译成中文，只输出中文翻译结果：\n The largest community building the future of LLM apps.")
# multi_prompt_chain.invoke("你是怎么理解java的面向对象的思想的？")
print(res)
```

## 5.记忆（memory）
当我们与 LLM 聊天时，它们无法记住上下文信息。

```python
from langchain_ollama import OllamaLLM
from langchain_core.prompts import ChatPromptTemplate

model = ChatOllama(model="qwen3:8b",url="http://localhost:11434")
prompt = ChatPromptTemplate.from_messages([
    ('system','你是一个AI助手'),
    ('user',"{question}")
])

chain = prompt | model

while True:
    question = input("请输入问题：")
    res = chain.invoke({"question": question})
    print(res)
    
```

LLM 固有缺陷，没有记忆。LLM 的本质是基于统计和概率来生成文本，对于每次请求，它们都将上下文视为独立事件。这意味着当你与 LLM 进行对话时，它不会记住你之前说过的话，这就导致了 LLM 有时表现得不够智能。<br>

这种“无记忆”属性使得 LLM 无法在长期对话中有效跟踪上下文，也无法积累历史信息。比如，当你在聊天过程中提到一个人名，后续再次提及该人时，LLM 可能会忘记你之前的描述。<br>

如果将已有信息放入到 memory 中，每次跟 LLM 对话时，把已有的信息丢给 LLM，那么 LLM 就能够正确回答。

目前业内解决 LLM 记忆问题就是采用了类似上图的方案，即：将每次的对话记录再次丢入到 Prompt 里，这样 LLM 每次对话时，就拥有了之前的历史对话信息。<br>

+ 在对话之前调取之前的历史消息。
+ 将历史消息填充到Prompt里。
+ 对话结束后，继续将历史消息保存到到memory记忆中。

但如果每次对话，都需要自己手动将本次对话信息继续加入到history信息中，那未免太繁琐。有没有轻松一些的方式呢？有，LangChain！LangChain 对记忆组件做了高度封装，开箱即用。<br>

在解决 LLM 的记忆问题时，有两种记忆方案，长期记忆和短期记忆。<br>
+ 短期记忆：基于内存的存储，容量有限，用于存储临时对话内容。<br>
+ 长期记忆：基于硬盘或者外部数据库等方式，容量较大，用于存储需要持久的信息。<br>

### 短期记忆 旧版

Langchain 提供了 ConversationBufferMemory 类，可以用来存储和管理对话。
ConversationBufferMemory 包含input变量和output变量，input代表人类输入，output代表 AI 输出。
每次往ConversationBufferMemory组件里存入对话信息时，都会存储到history的变量里。

```python
from langchain_classic.memory import ConversationBufferMemory

memory = ConversationBufferMemory(return_messages=True)#return_messages=True 返回消息列表
memory.save_context({"input": "我的名字叫张三"}, {"output": "你好，张三"})#存入对话

res1 = memory.load_memory_variables({})#读取对话
print(res1)

memory.save_context({"input": "我是一名 IT 程序员"}, {"output": "好的，我知道了"})
res2 = memory.load_memory_variables({})

```

在 Langchain 中，MessagesPlaceholder是一个占位符，用于在对话模板中动态插入上下文信息。MessagesPlaceholder 接收的是 消息列表（list），不能传单条消息；列表里每一项会被 LangChain 转成 BaseMessage。它可以帮助我们灵活地管理对话内容，确保 LLM 能够使用最上下文来生成响应。<br>

采用ConversationChain对话链结合PromptTemplate和MessagesPlaceholder，几行代码就可以轻松让 LLM 拥有短时记忆。<br>

```python
# 利用 MessagesPlaceholder 手动添加 history
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "你是一个乐于助人的助手。"),
        MessagesPlaceholder(variable_name="history"),
        ("human", "{user_input}"),
    ]
)
user_input = "你知道我的名字吗？"
history = memory.load_memory_variables({})["history"]
print(history)

chain = prompt | model
res = chain.invoke({"user_input": user_input, "history": history})
print(res)
```

我们利用 LangChain 的ConversationChain对话链，自动添加history的方式添加临时记忆，无需手动添加。一个链实际上就是将一部分繁琐的小功能做了高度封装，这样多个链就可以组合形成易用的强大功能。这里链的优势一下子就体现出来了：

```python
#利用 ConversationChain 自动添加 history
from langchain_classic.chains import ConversationChain
from langchain_classic.memory import ConversationBufferMemory

memory = ConversationBufferMemory(return_messages=True)
chain = ConversationChain(llm=model, memory=memory)
res = chain.invoke({"input": "你好，我的名字是ryan，我是一名程序员。"})
print(res['response'])
```

```python
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_openai import ChatOpenAI

# 1. 用字典模拟不同用户/会话的历史记录
store = {}

def get_session_history(session_id: str):
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()
    return store[session_id]

# 2. prompt 里用 MessagesPlaceholder 放历史消息
prompt = ChatPromptTemplate.from_messages([
    ("system", "你是一个有帮助的AI助手。"),
    MessagesPlaceholder(variable_name="history"),
    ("human", "{input}"),
])

# 3. 普通链
llm = ChatOpenAI(
    model_name="qwen3.6-flash",
    api_key="",
    base_url="https://llm-0vksjtr7rk6p888v.cn-beijing.maas.aliyuncs.com/compatible-mode/v1"  
)
chain = prompt | llm

# 4. 包装成带记忆的 chain
chain_with_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="input",
    history_messages_key="history",
)

# 5. 调用时必须传 session_id
resp1 = chain_with_history.invoke(
    {"input": "我叫张三"},
    config={"configurable": {"session_id": "user-001"}}
)

print(resp1.content)0

resp2 = chain_with_history.invoke(
    {"input": "我叫什么？"},
    config={"configurable": {"session_id": "user-001"}}
)

print(resp2.content)
```

### 长期记忆
短期记忆在会话关闭或者服务器重启后，就会丢失。如果想长期记住对话信息，只能采用长期记忆组件。

LangChain 支持多种长期记忆组件，比如Elasticsearch、MongoDB、Redis等，下面以Redis为例，演示如何使用长期记忆。

安装redis<br>
docker run -d --name redis -p 6379:6379 redis:latest

本地<br>
pip install redis

测试<br>
redis-cli -h 127.0.0.1 -p 6379 ping

## 6.AI Agent
### Agent的概念<br>
AI Agent，经常被翻译为：智能体或代理。<br>
一句话总结，AI Agent就是一个有着聪明大脑而且能够感知外部环境并采取行动的智能系统。<br>

Agent 三要素<br>
<br>

我们可以把它想象成一个能思考和行动的人，而大型语言模型（LLM）就是这个人的“大脑”。通过这个大脑，再加上一些能够感知外部世界和执行任务的部件，AI Agent就变成了一个有“智慧”的机器人。<br>

    <source src="./agent.mp4" type="video/mp4">
</video>
<br>

要让AI Agent充分利用它的“大脑”和各种组件，需要一种协调机制。ReAct机制就是常用的协调机制。通过ReAct机制，AI Agent能够结合外部环境和行动组件，完成复杂的任务。<br>

为什么我们需要AI Agent呢？其实说到底是因为单一的模型对我们来说作用不大，我们需要的是一个具备智能的复杂系统。只有复杂系统才能真正的应到到实际生产工作中。<br>


Agent代理的工作原理可以概括为以下几个步骤：

+ 接收用户输入：Agent首先接收用户的输入，这可以是一个问题、一个请求或一个指令。
+ 分析并决策：接着，Agent利用语言模型对用户输入进行分析，并根据分析结果决定要采取的操作。这个决策过程可能涉及到对多个可能操作的评估，以及根据上下文选择最合适的操作。
+ 调用工具并执行：一旦决定了要采取的操作，Agent就会调用相应的工具来执行该操作。这些工具可以是LangChain提供的内置工具，也可以是开发者自定义的工具。
+ 返回结果并更新状态：执行完操作后，Agent将结果返回给用户，并更新其内部状态，以便处理后续的用户输入。

```python
from langchain_ollama import ChatOllama
from langchain.tools import tool
from langchain.agents import create_tool_calling_agent, AgentExecutor
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_classic.memory import ConversationBufferMemory
# 1. 定义工具
@tool
def get_weather(city: str) -> str:
    """获取城市天气信息"""
    return f"{city}的天气是晴天"

@tool
def calculate(expression: str) -> str:
    """计算数学表达式"""
    try:
        result = eval(expression)
        return f"计算结果: {result}"
    except:
        return "计算错误"

# 2. 初始化支持 function call 的模型
model = ChatOllama(model="qwen2.5:7b", base_url="http://localhost:11434")
memory = ConversationBufferMemory(return_messages=True,memory_key="chat_history")
# 3. 定义提示模板
prompt = ChatPromptTemplate.from_messages([
    ("system", "你是一个有用的助手"),
    MessagesPlaceholder(variable_name="chat_history"),
    ("user", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad")
])

# 4. 创建工具调用 Agent
tools = [get_weather, calculate]
agent = create_tool_calling_agent(model, tools, prompt)

#支持 Function Calling 的模型可以在一次响应中同时调用多个工具
# → AgentExecutor 并行执行两个工具
# → 两个结果一起返回给模型
# → 模型给出最终答案

# 5. 创建 Agent Executor
agent_executor = AgentExecutor(
    agent=agent, 
    tools=tools, 
    verbose=True,
    handle_parsing_errors=True,#工具调用报错时不抛异常，而是返回一个错误提示让 agent 继续处理
    memory=memory
)

# 6. 使用
result = agent_executor.invoke({"input": "北京天气如何？"})
print(result["output"])
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.tools import tool
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_ollama import ChatOllama
from langchain_classic.agents import create_tool_calling_agent, AgentExecutor


# 1. 模型（注意：必须用 ChatOllama，不能用 OllamaLLM）
model = ChatOllama(model="qwen3:14b",base_url="http://192.168.1.105:11434")


# 2. 定义工具
@tool
def get_weather(city: str) -> str:
    """获取某城市的天气"""
    return f"{city}今天天气晴，气温25度"

@tool
def calculator(expression: str) -> str:
    """计算数学表达式"""
    try:
        return str(eval(expression))
    except Exception as e:
        return f"计算失败：{e}"

tools = [get_weather, calculator]


# 3. Prompt 里必须包含 chat_history 和 agent_scratchpad
prompt = ChatPromptTemplate.from_messages([
    ("system", "你是一个智能助手，可以使用工具帮助用户。"),
    MessagesPlaceholder(variable_name="chat_history"),   # 历史消息
    ("human", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad"),  #存放工具结果的
])


# 4. 创建 agent
agent = create_tool_calling_agent(model, tools, prompt)


# 5. 创建 AgentExecutor
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)


# 6. 历史存储
store = {}

def get_session_history(session_id: str):
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()
    return store[session_id]


# 7. 包装成带历史的 executor
agent_with_history = RunnableWithMessageHistory(
    agent_executor,
    get_session_history,
    input_messages_key="input",
    history_messages_key="chat_history",
    handle_parsing_errors=True,#工具调用报错时不抛异常，而是返回一个错误提示让 agent 继续处理
)


# 8. 调用
res = agent_with_history.invoke(
    {"input": "北京天气怎么样？"},
    config={"configurable": {"session_id": "user_001"}},
)
print(res["output"])

res = agent_with_history.invoke(
    {"input": "我刚才问的什么？"},
    config={"configurable": {"session_id": "user_001"}},
)
print(res["output"])
```

#### 不支持functioncall的模型调用工具（了解即可）

```python
from langchain_core.prompts import ChatPromptTemplate,PromptTemplate, prompt
from langchain_core.messages import SystemMessage,HumanMessage
from langchain_ollama import ChatOllama
from langchain_classic.globals import set_debug, set_verbose
# 启用调试模式
set_debug(True)
set_verbose(True)

#不支持functioncall 的模型使用工具

import json
import re
from datetime import datetime

from langchain_ollama import ChatOllama


llm = ChatOllama(model="qwen3:8b", base_url="http://localhost:11434")

SYSTEM_PROMPT = """
你是一个可以借助外部工具完成任务的助手。

可用工具：
1. get_time()
作用：获取当前时间

规则：
1. 如果需要调用工具，必须严格输出：
<tool_call>
{"name":"工具名","arguments":{}}
</tool_call>

2. 不要编造工具结果
3. 每次只允许调用一个工具
4. 如果不需要工具，直接正常回答
""".strip()

# =========================
# 1. 调用大模型的函数
# =========================
def call_llm(messages):
    response = llm.invoke(messages)
    return response.content


# =========================
# 2. 定义真实工具
# =========================
def get_time():
    return datetime.now().strftime("%Y-%m-%d %H:%M:%S")


TOOLS = {
    "get_time": get_time
}


# =========================
# 3. 工具调用解析
# =========================
def parse_tool_call(text):
    """
    从模型输出中提取：
    <tool_call>
    {...json...}
    </tool_call>
    """
    pattern = r"<tool_call>\s*(\{.*?\})\s*</tool_call>"
    match = re.search(pattern, text, re.DOTALL)
    if not match:
        return None

    try:
        data = json.loads(match.group(1))
        return data
    except json.JSONDecodeError:
        return None

def extract_tool_result(text):
    pattern = r"工具返回结果[:：]\s*(.*)"
    match = re.search(pattern, text, re.DOTALL)
    return match.group(1).strip() if match else "未知结果"


# =========================
# 4. 执行工具
# =========================
def run_tool(name, arguments):
    if name not in TOOLS:
        return f"错误：未知工具 {name}"

    try:
        result = TOOLS[name](**arguments)
        return result
    except Exception as e:
        return f"工具执行失败: {e}"


# =========================
# 5. 主流程：提示词模拟 tool calling
# =========================

def chat(user_input):
    # 第一轮：让模型决定是否调用工具
    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": user_input},
    ]

    llm_output = call_llm(messages)
    print("模型第一轮输出：")
    print(llm_output)
    print("-" * 40)

    tool_call = parse_tool_call(llm_output)

    # 如果没有工具调用，就直接返回
    if not tool_call:
        return llm_output

    # 执行工具
    tool_name = tool_call["name"]
    arguments = tool_call.get("arguments", {})
    tool_result = run_tool(tool_name, arguments)

    # 第二轮：把工具结果发回给模型，让它组织自然语言
    messages.append({"role": "assistant", "content": llm_output})
    messages.append({
        "role": "user",
        "content": f"工具返回结果: {tool_result}\n请根据工具结果回答用户。"
    })

    final_answer = call_llm(messages)
    return final_answer


# =========================
# 6. 测试
# =========================
if __name__ == "__main__":
    user_question = "现在几点了？"
    answer = chat(user_question)

    print("最终回答：")
    print(answer)
```

#### React 推理模式

ReAct 是 Reasoning + Acting 的缩写，是一种让大模型通过"思考-行动-观察"循环来解决复杂问题的 Agent 模式。


核心思想
思考(Thought) → 行动(Action) → 观察(Observation) → 思考 → ... → 最终答案
模型不是一次性给出答案，而是像人一样：
Thought：分析当前情况，决定下一步做什么
Action：调用某个工具（搜索、计算、查数据库等）
Observation：观察工具返回的结果
重复以上步骤，直到得出最终答案


支持不具备function call能力的模型

```python
react_prompt = PromptTemplate.from_template("""
你是一个有用的助手。你可以使用以下工具：

{tools}

工具名称: {tool_names}

使用以下格式回答：

Question: 用户的问题
Thought: 你应该思考要做什么
Action: 要使用的工具，必须是 [{tool_names}] 中的一个  #
Action Input: 工具的输入
Observation: 工具返回的结果
... (可以重复 Thought/Action/Action Input/Observation 多次)
Thought: 我现在知道最终答案了
Final Answer: 对原始问题的最终答案
对话历史：
{chat_history}

开始!

Question: {input}
Thought: {agent_scratchpad}
""")


'''
| 变量 | 作用 | 是否必须 |
|------|------|----------|
| `{tools}` | 工具的详细描述（名称+说明） | 必须 |
| `{tool_names}` | 工具名称列表 | 必须 |
| `{input}` | 用户输入 | 必须 |
| `{agent_scratchpad}` | Agent 中间思考步骤 | 必须 |
| `{chat_history}` | 历史对话（使用 memory 时需要） | 使用 memory 时必须 |
这些变量会在运行时由 `AgentExecutor` **自动填充**，不需要手动传入
'''
```

```python
from langchain_ollama import ChatOllama
from langchain.tools import tool
from langchain_classic.agents import create_react_agent, AgentExecutor
from langchain_core.prompts import PromptTemplate
from langchain_classic.memory import ConversationBufferMemory


# 定义工具
@tool
def get_weather(city: str) -> str:
    """获取城市天气信息"""
    weather_data = {
        "北京": "晴天，25度",
        "上海": "多云，22度",
        "深圳": "雨天，28度"
    }
    return weather_data.get(city, f"{city}天气未知")

@tool
def calculate(expression: str) -> str:
    """计算数学表达式"""
    try:
        result = eval(expression)
        return f"计算结果: {result}"
    except:
        return "计算错误"

# 创建模型
model = ChatOllama(model="qwen3:14b", base_url="http://192.168.1.105:11434")

# ReAct 提示词模板
react_prompt = PromptTemplate.from_template("""
你是一个有用的助手。你可以使用以下工具：

{tools}

工具名称: {tool_names}

使用以下格式回答：

Question: 用户的问题
Thought: 你应该思考要做什么
Action: 要使用的工具，必须是 [{tool_names}] 中的一个  #
Action Input: 工具的输入
Observation: 工具返回的结果
... (可以重复 Thought/Action/Action Input/Observation 多次)
Thought: 我现在知道最终答案了
Final Answer: 对原始问题的最终答案
对话历史：
{chat_history}

开始!

Question: {input}
Thought: {agent_scratchpad}
""")


# 创建 ReAct Agent
tools = [get_weather, calculate]
agent = create_react_agent(model, tools, react_prompt)

memory = ConversationBufferMemory(
    memory_key="chat_history",
    return_messages=False,
    input_key="input",
    output_key="output"
)
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True,memory=memory,
                                max_iterations=5,handle_parsing_errors=True)
# 作用：执行引擎
# 能力：运行 agent + 执行工具 + 管理循环
# 依赖：需要一个 agent 来做决策
result = agent_executor.invoke({"input": "北京天气如何温度是多少？，你需要讲得到的温度+20"})
print(result["output"])

```

```python
from langchain import memory
from langchain_core.prompts import ChatPromptTemplate,PromptTemplate,MessagesPlaceholder
from langchain_core.messages import SystemMessage,HumanMessage
# from langchain_ollama.llms import OllamaLLM
from langchain_classic.globals import set_debug, set_verbose
from langchain_ollama import ChatOllama
from langchain_core.tools import tool
from langchain_classic.memory import ConversationBufferMemory
from langchain_classic.agents import initialize_agent,AgentType
# # 启用调试模式
# # set_debug(True)
# # set_verbose(True)

llm = ChatOllama(model="qwen3:8b",base_url="http://localhost:11434")  #不能使用deepseek  



memory = ConversationBufferMemory(return_messages=True,memory_key="chat_history")

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers."""
    return a * b
agent = initialize_agent(
    tools=[multiply],
    llm=llm,
    agent=AgentType.STRUCTURED_CHAT_ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True,
    handle_parsing_errors=True, #自动处理ai输出错误的格式
    agent_kwargs={
        # 自定义系统提示
        "system_message": "你是一个专业的旅游规划师，擅长为用户推荐旅游景点和制定行程计划。",
        
        # 插入历史对话
        "extra_prompt_messages": [
            MessagesPlaceholder(variable_name="chat_history")
        ],
    },
    memory=memory,
    max_iterations=5 #限制react的步数
    )


res = agent.invoke({"input": "355222*5521222等于多少"})
print(memory.load_memory_variables({}))
# print(res)
```
